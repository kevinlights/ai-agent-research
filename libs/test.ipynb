```python
from autogen_core.models import ModelFamily, UserMessage, ChatCompletionClient, SystemMessage
from autogen_ext.models.openai import OpenAIChatCompletionClient
from typing import Mapping
```

```python
client = OpenAIChatCompletionClient(
    model="qwen3:4b",
    api_key="ollama",
    base_url="http://localhost:11434/v1",
    model_info={
        "json_output": False,
        "vision": False,
        "function_calling": True,
        "family": ModelFamily.ANY,
    },
)


stream = client.create_stream(
    messages=[
        SystemMessage(content="/no_think"),
        UserMessage(content="请介绍一下 qwen3", source="user"),
    ],
    extra_create_args={"stream_options": {"include_usage": True}},
)
async for response in stream:
    if isinstance(response, str):
        # A partial response is a string.
        print(response, flush=True, end="")
    else:
        # The last response is a CreateResult object with the complete message.
        print("\\n\\n------------\\n")
        print("The complete response:", flush=True)
        print(response.content, flush=True)

# Close the client when done.
await client.close()
```
